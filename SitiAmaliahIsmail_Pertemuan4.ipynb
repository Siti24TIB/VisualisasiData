```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
#Table Data
path=r"Siti.xlsx"
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
#Table Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
#Grafik Garis
datafrq.plot()

plt.title('Diagram Garis Grade')
plt.show()
```

```python
#Grafik Batang
datafrq.plot(kind="bar")

plt.title('Diagram Batang Grade')
plt.show()
```

```python
#Pie Chart
datafrq.plot(kind='pie',y='Frekuensi',autopct='%1.0f%%',legend=False)

plt.ylabel('')
plt.title('Diagram Pie Frekuensi')
plt.show()
```

```python
#Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors="coerce")
dt = dataraw["Final Score"]
stats = dt.describe()
stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["Variance"] = dt.var()
stats["range"] = dt.max() - dt.min()
stats["skewness"] = dt.skew()
stats["kurtosis"] = dt.kurtosis()
print(stats)
```
